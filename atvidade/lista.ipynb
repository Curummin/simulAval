# Exercício 2

```python
espaco_amostral = []

# 0 = cara
# 1 = coroa
for i in range(0,2):
  for j in range(0,2):
    for k in range(0,2):
      for l in range(0,2):
        for m in range(0,2):
          espaco_amostral.append((i,j,k,l,m))
print(f"Espaço amostral: {espaco_amostral}")

valores_X = []  #soma

for i,j,k,l,m in espaco_amostral:
    valores_X.append(i+j+k+l+m)

total_eventos = len(espaco_amostral)

# (a) P(X > 3) = P(X = 4) + P(X = 5)
contagem_X4 = valores_X.count(4)
contagem_X5 = valores_X.count(5)
P_X_maior_3 = (contagem_X4 + contagem_X5) / total_eventos

# (b) P(X = 2)
contagem_X2 = valores_X.count(2)
P_X_igual_2 = contagem_X2 / total_eventos

# (c) E(X) = valor esperado
E_X = sum(valores_X) / total_eventos

# Resultados
print(f"Total de eventos: {total_eventos}")
print(f"\n(a) P(X > 3) = {P_X_maior_3}")
print(f"    ({contagem_X4} eventos com X=4 + {contagem_X5} eventos com X=5) / {total_eventos}")
print(f"    = {P_X_maior_3:.4f} ou {P_X_maior_3 * 100:.1f}%")

print(f"\n(b) P(X = 2) = {P_X_igual_2}")
print(f"    ({contagem_X2} eventos) / {total_eventos}")
print(f"    = {P_X_igual_2:.4f} ou {P_X_igual_2 * 100:.1f}%")

print(f"\n(c) E(X) = {E_X}")
print(f"    Soma total de caras: {sum(valores_X)} / {total_eventos}")
print(f"    = {E_X:.2f} caras em média")

```

# Exercício 3

```python
# Representação das bolas: 0,1,2 são Vermelhas | 3,4 são Azuis
urnas = [0, 1, 2, 3, 4]
espaco_amostral = []

for i in urnas:
    for j in urnas:
        if i != j:
            espaco_amostral.append((i, j))

print("Espaco amostral:")
print(espaco_amostral)

valores_X = []
for i, j in espaco_amostral:
    contagem_vermelhas = 0
    if i < 3: contagem_vermelhas += 1
    if j < 3: contagem_vermelhas += 1
    valores_X.append(contagem_vermelhas)

print("\nValores de X = soma:")
print(valores_X)

contagem_frequencia = [0] * 3

for x in valores_X:
    contagem_frequencia[x] += 1

print("\nDistribuição de Probabilidades de X:")
probabilidades = []

for k in range(3):
    prob = contagem_frequencia[k] / len(espaco_amostral)
    probabilidades.append(prob)
    print(f"P(X = {k}): {prob*100:.1f}%")

valores_possiveis = [0, 1, 2]
esperanca = np.average(valores_possiveis, weights=probabilidades)

print(f"\nEsperanca: {esperanca:.2f}")
```

# Exercício 5
Neste exercício 5, usamos o valor de n = 5 para teste, mas qualquer valor poderia ser usado.

Então, calculamos todas as permutações possíveis, para podermos determinar o espaço amostral e todas as situações em que uma carta se mantém no mesmo lugar.
Depois disso, podemos calcular a probabilidade da quantidade de cada carta estar ordenada, e a esperança matemática.

O que é interessante desse problema, é que independente do número de cartas que sejam embaralhadas, a esperança sempre será 1.

```python
import numpy as np

n = 7

cartoes_originais = list(range(1, n + 1))
espaco_amostral = list(itertools.permutations(cartoes_originais))

print(f"\nNúmero total de permutações: {len(espaco_amostral)}")

valores_X_lista = []

for permutacao in espaco_amostral:
    coincidencias = 0
    for i in range(n):
        if permutacao[i] == cartoes_originais[i]:
            coincidencias += 1
    valores_X_lista.append(coincidencias)

contagem_frequencia = [0] * (n + 1)
for x in valores_X_lista:
    contagem_frequencia[x] += 1

print("\ncontagem frequencia:")
print(contagem_frequencia)

probabilidades = []
for k in range(n + 1):
    prob = contagem_frequencia[k] / len(espaco_amostral)
    probabilidades.append(prob)

print("\nprobabilidades:")
print(probabilidades)


valores_possiveis = list(range(n + 1))
esperanca = np.average(valores_possiveis, weights=probabilidades)

print(f"\nEsperança Matemática E(X) para n={n}: {esperanca:.2f}")
```

```python
# Exercício 6
import math
import matplotlib.pyplot as plt
import numpy as np
# cara = p      #probabilidade de sucesso
# coroa = 1-p   #probabilidade de fracasso
# n lançamentos
# X = numero de cara obtidas

def P(k, n, p):
    if k < 0 or k > n:
      return 0

    coefBinomial = math.factorial(n) // (math.factorial(k) * math.factorial(n-k)) 
    probabilidade = coefBinomial * (p ** k) * ((1-p) ** (n-k))
    return probabilidade


def E(p,n):
  return p * n

def V(p,n):
  return n * p * (1-p)

p_sucesso = float(input("Entre com a probabilidade de sucesso (0.0 <= p <= 1.0): "))
n = int(input("Entre com o numero de lances: "))
k = int(input("Entre com o numero de sucessos para se calcular a probabilidade: "))

print(f"Probabilidade de Cara: {P(k, n, p_sucesso)}")
print(f"Esperança E(X): {E(p_sucesso,n)}")
print(f"Variância V(X): {V(p_sucesso, n):.2f}")

# 1. Gerar todos os valores possíveis de k (de 0 até n)
eixo_x = np.arange(0, n + 1)

# 2. Calcular a probabilidade para cada valor de k usando a função P
eixo_y = [P(i, n, p_sucesso) for i in eixo_x]

# 3. Criar o gráfico
plt.figure(figsize=(10, 6))
plt.bar(eixo_x, eixo_y, color='skyblue', edgecolor='navy', alpha=0.7)
plt.title(f'Distribuição Binomial (n={n}, p={p_sucesso})', fontsize=14)
plt.xlabel('Número de Caras (k)', fontsize=12)
plt.ylabel('Probabilidade P(X = k)', fontsize=12)
plt.xticks(eixo_x)  
plt.grid(axis='y', linestyle='--', alpha=0.6)
for i in range(len(eixo_x)):
    plt.text(eixo_x[i], eixo_y[i], f'{eixo_y[i]:.3f}', ha='center', va='bottom', fontsize=9)

plt.show()
```

```python
# Exercício 10

import numpy as np
import matplotlib.pyplot as plt

x = [0 , 1 , 2 , 3 ]
p_x = [0.1 , 0.3, 0.4, 0.2]

# (a) Função de distribuição acumulada (CDF)
F_x = np.cumsum(p_x)

print("Valores de x:", x)
print("p(x):", p_x)
print("F(x):", F_x)

# (b) Cálculos pedidos
# i) F(1)
F_1 = F_x[1]

# ii) F(2)
F_2 = F_x[2]

# iii) P(1 < X ≤ 3)
P_intervalo = np.sum(p_x[2:])

print("\nResultados:")
print("F(1) =", F_1)
print("F(2) =", F_2)
print("P(1 < X ≤ 3) =", P_intervalo)

# Plot da CDF
plt.plot(x , F_x , 'bo', markersize =5) # pontos azuis
# Plota as linhas horizontais entre os pontos
for i in range (1 , len( x)):
  plt.hlines(F_x [i -1] , x [i -1] , x [i ], colors ='b', linestyles ='solid')

for i in range (1 , len( x) - 1) :
  plt.vlines (x [i] , F_x [i -1] , F_x [i ], colors ='b', linestyles ='dotted')

plt.title (" Função de Distribuição Acumulada de X")
plt.xlabel ("x")
plt.ylabel (r" $F_X (x)$")
plt.grid ( True )
plt.show ()
```
